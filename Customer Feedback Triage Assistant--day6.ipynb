An e-commerce company receives thousands of customer complaints. It needs a system that can automatically analyse each complaint and help the support team decide what to handle first.

```python
feedback_data = [
    {"id": "FB001", "text": "My refund has not arrived after 10 days."},
    {"id": "FB002", "text": "Excellent product and very fast delivery."},
    {"id": "FB003", "text": "The package arrived damaged and customer support did not respond."},
    {"id": "FB004", "text": "I was charged twice for the same order."},
    {"id": "FB005", "text": "Delivery was late, but the product is good."},
    {"id": "FB006", "text": "The product is okay. Nothing special."}
]
```

```python
for item in feedback_data:
    print(item["id"], "-", item["text"])
```

```python
def analyse_feedback(text):

    text = text.lower()

    # Category
    if "refund" in text or "charged" in text:
        category = "Billing"
    elif "delivery" in text or "package" in text:
        category = "Delivery"
    else:
        category = "Other"

    # Sentiment
    if "damaged" in text or "not arrived" in text or "charged twice" in text:
        sentiment = "Negative"
    elif "excellent" in text or "good" in text:
        sentiment = "Positive"
    else:
        sentiment = "Neutral"

    # Urgency
    if "refund" in text or "charged twice" in text or "damaged" in text:
        urgency = "High"
    elif "late delivery" in text or "no response" in text:
        urgency = "Medium"
    else:
        urgency = "Low"
        
    # After calculating urgency

    if "refund" in text:
        reason = "Customer is waiting for a refund"

    elif "charged twice" in text:
        reason = "Customer reports duplicate payment"

    elif "damaged" in text:
        reason = "Product or package arrived damaged"

    elif "late" in text:
        reason = "Delivery was delayed"

    elif "no response" in text:
        reason = "Customer support did not respond"

    else:
        reason = "No major issue detected"

    return {
        "category": category,
        "sentiment": sentiment,
        "urgency": urgency,
        "reason": reason

    }
```

```python
for item in feedback_data:
    result = analyse_feedback(item["text"])
    print(item["id"], result)
```

```python
results = []

for item in feedback_data:
    result = analyse_feedback(item["text"])

    results.append({
        "id": item["id"],
        **result
    })

for record in results:
    print(record)
```

```python
import pandas as pd

df = pd.DataFrame(results)

print(df)
```

```python
# Count complaints by category
df["category"].value_counts().groupby(df["category"])
```
