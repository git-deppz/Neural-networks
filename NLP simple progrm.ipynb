##So CountVectorizer gives the number of times each word appears and TfidfVectorizer gives the weight of each.
We fit the TF-IDF vectorizer only on training data to prevent data leakage, and use the same fitted vectorizer to transform unseen test data."

```python
reviews = [
    "The product is excellent",
    "I love this product",
    "Very happy with the purchase",
    "The quality is amazing",
    "The product is terrible",
    "Very poor quality",
    "I hate this product",
    "Delivery was disappointing"
]

labels = [
    1, 1, 1, 1,
    0, 0, 0, 0
]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    reviews,
    labels,
    test_size=0.25,
    random_state=42,
    stratify=labels
)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(ngram_range=(1, 2))
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)
```

```python
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()

model.fit(X_train_tfidf, y_train)
```

```python
predictions = model.predict(X_test_tfidf)

print(predictions)
print(y_test)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

print("Accuracy:", accuracy_score(y_test, predictions))
print(confusion_matrix(y_test, predictions))
print(classification_report(y_test, predictions, zero_division=0))
```
